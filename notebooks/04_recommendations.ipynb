```python
severity_counts = df_incidents['severity'].value_counts()
sla_breaches = df_incidents['sla_breach'].sum() if 'sla_breach' in df_incidents.columns else 'N/A'

print(f"1. Всего проанализировано инцидентов: {len(df_incidents)}")
print(f"2. Распределение критичности инцидентов:\n{severity_counts}")
print(f"3. Общее количество нарушений SLA: {sla_breaches}")
```
